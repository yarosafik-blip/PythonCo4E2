## Exercise 1: List basics

1. Create `fruits = ["apple", "banana", "cherry", "mango"]`.
2. Print the first item, last item, and length of the list.
3. Change `"banana"` to `"orange"` and print the updated list.

```python
fruits = ["apple", "banana", "cherry", "mango"]
print("First item:", fruits[0])
print("Last item:", fruits[-1])
print("Length of list:", len(fruits))
fruits[1] = "orange"
print("Updated list:", fruits)
```

## Exercise 2: Add and remove

Using the same `fruits` list:

1. Add `"grape"` using `append()`.
2. Insert `"kiwi"` at index 1 using `insert()`.
3. Remove `"cherry"` using `remove()`.
4. Remove the last item using `pop()`.
5. Print the list after each change.

```python
fruits.append("grape")
print("After append('grape'):", fruits)
fruits.insert(1, "kiwi")
print("After insert(1, 'kiwi'):", fruits)
fruits.remove("cherry")
print("After remove('cherry'):", fruits)
popped_item = fruits.pop()
print(f"After pop() (removed '{popped_item}'):", fruits)
```

## Exercise 3: Loop through a list

1. Create `scores = [75, 88, 92, 60, 81]`.
2. Use a `for` loop to print each score.
3. In another loop, print only scores greater than or equal to 80.
4. Calculate and print the total score.

```python
scores = [75, 88, 92, 60, 81]

print("All scores:")
for score in scores:
    print(score)

print("\nScores >= 80:")
for score in scores:
    if score >= 80:
        print(score)

total_score = sum(scores)
print("\nTotal score:", total_score)
```

## Exercise 4: Comprehension and sort

1. From `scores = [75, 88, 92, 60, 81]`, create `high_scores` with only scores >= 80 using list comprehension.
2. Sort `high_scores` in ascending order.
3. Sort it again in descending order.
4. Print all results.

```python
scores = [75, 88, 92, 60, 81]
high_scores = [score for score in scores if score >= 80]
print("High scores (filtered):", high_scores)

high_scores.sort()
print("Ascending order:", high_scores)

high_scores.sort(reverse=True)
print("Descending order:", high_scores)
```

## Exercise 5: Copy and join lists

1. Create `list_a = [1, 2, 3]` and `list_b = [4, 5, 6]`.
2. Make a copy of `list_a` in a new variable.
3. Join `list_a` and `list_b` into one list in two ways:
   - using `+`
   - using `extend()`
4. Print all lists.

```python
list_a = [1, 2, 3]
list_b = [4, 5, 6]

list_a_copy = list_a.copy()

joined_plus = list_a + list_b

joined_extend = list_a.copy()
joined_extend.extend(list_b)

print("Original list_a:", list_a)
print("Original list_b:", list_b)
print("Copy of list_a:", list_a_copy)
print("Joined using '+':", joined_plus)
print("Joined using extend():", joined_extend)
```
