## Exercise 1: Nested list processing

- Create a 2D list (3 rows x 4 cols) of numbers.
- Print row sums and grand total.
- Print the maximum value from each row.

```python
matrix = [
    [10, 25, 3, 14],
    [5, 18, 92, 7],
    [33, 41, 12, 60]
]

grand_total = 0
for idx, row in enumerate(matrix, start=1):
    row_sum = sum(row)
    grand_total += row_sum
    print(f"Row {idx}: {row} | Sum: {row_sum} | Max: {max(row)}")

print("Grand total:", grand_total)
```

## Exercise 2: Comprehension with condition

- From numbers 1..30, build a list of numbers divisible by 3 but not by 2.
- Then build a second list of their squares using comprehension.
- Print both lists.

```python
divisible_by_3_not_2 = [n for n in range(1, 31) if n % 3 == 0 and n % 2 != 0]
squares = [n**2 for n in divisible_by_3_not_2]

print("Numbers divisible by 3 but not 2 (1..30):", divisible_by_3_not_2)
print("Squares of those numbers:", squares)
```

## Exercise 3: Custom sorting

- Create a list of lists: `[name, score]` for at least 5 students.
- Write a helper function that returns a sort key `[-score, name]`.
- Sort descending by score, then ascending by name for ties, using `sort(key=...)`.
- Print the sorted result.

```python
students = [
    ["Alice", 85],
    ["Bob", 92],
    ["Charlie", 85],
    ["Diana", 78],
    ["Ethan", 92]
]

def sort_key(student):
    return [-student[1], student[0]]

students.sort(key=sort_key)
print("Sorted students (score desc, name asc):")
for student in students:
    print(student)
```

## Exercise 4: Duplicate removal (order preserved)

- Given a list with duplicates, create a new list without duplicates while preserving original order.
- Use a loop and a membership check (`not in`) before appending.
- Print original and cleaned lists.

```python
original_list = [4, 2, 7, 2, 4, 9, 1, 7, 3, 1]

cleaned_list = []
for item in original_list:
    if item not in cleaned_list:
        cleaned_list.append(item)

print("Original list:", original_list)
print("Cleaned list (order preserved):", cleaned_list)
```
