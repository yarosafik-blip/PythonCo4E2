## Exercise 1: Boolean basics

1. Create variables `x = 10`, `y = 0`, and `text = ""`.
2. Print `bool(x)`, `bool(y)`, and `bool(text)`.
3. Write one `if` statement that prints `"x is truthy"` when `x` is True-like.

```python
x = 10
y = 0
text = ""
print('bool(x):', bool(x))
print('bool(y):', bool(y))
print('bool(text):', bool(text))
if x:
    print('x is truthy')
```

## Exercise 2: Comparison operators

Given `a = 12` and `b = 7`, print the result of:

- `a > b`
- `a < b`
- `a == b`
- `a != b`
- `a >= 12`

```python
a = 12
b = 7
print('a > b:', a > b)
print('a < b:', a < b)
print('a == b:', a == b)
print('a != b:', a != b)
print('a >= 12:', a >= 12)
```

## Exercise 3: Logical operators

1. Create `age = 20` and `has_id = True`.
2. Use `and` to check if someone can enter: age >= 18 and has_id is True.
3. Use `or` to check if `age < 18` or `has_id` is False.
4. Use `not` with `has_id` and print the result.

```python
age = 20
has_id = True
print('Can enter:', age >= 18 and has_id is True)
print('Underage or missing ID:', age < 18 or has_id is False)
print('Does not have ID:', not has_id)
```

## Exercise 4: Identity and membership

1. Create two lists:
   - `list1 = [1, 2, 3]`
   - `list2 = list1`
2. Print `list1 is list2` and `list1 is not list2`.
3. Check membership in `list1`:
   - `2 in list1`
   - `5 not in list1`

```python
list1 = [1, 2, 3]
list2 = list1
print('Same object:', list1 is list2)
print('Different objects:', list1 is not list2)
print('2 in list1:', 2 in list1)
print('5 not in list1:', 5 not in list1)
```

## Exercise 5: Arithmetic and assignment

1. Create `n = 10`.
2. Update `n` using these operators in order and print after each step:
   - `n += 5`
   - `n -= 3`
   - `n *= 2`
   - `n /= 4`
3. Print `n % 3` and `n ** 2`.

```python
n = 10
n += 5
print('After += 5:', n)
n -= 3
print('After -= 3:', n)
n *= 2
print('After *= 2:', n)
n /= 4
print('After /= 4:', n)
print('n % 3:', n % 3)
print('n ** 2:', n ** 2)
```
