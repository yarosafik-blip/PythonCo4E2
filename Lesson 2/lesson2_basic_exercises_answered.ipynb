## Exercise 1: Numeric types

1. Create three variables:
   - `a` as an integer
   - `b` as a float
   - `c` as a complex number
2. Print each variable.
3. Print `type(...)` for each variable.
4. Create a float in scientific notation and print it.

````python
a = 10
b = 3.5
c = 2 + 4j
for value in (a, b, c):
    print(value, type(value))
scientific = 1.25e3
print('Scientific notation:', scientific)
````

## Exercise 2: Casting values

1. Convert `19` to float and store it in `x`.
2. Convert `7.95` to int and store it in `y`.
3. Convert `"42"` to int and store it in `z`.
4. Convert `25` to string and combine it into: `"Age: 25"`.
5. Print all results and their types.

````python
x = float(19)
y = int(7.95)
z = int("42")
age_text = "Age: " + str(25)
for result in (x, y, z, age_text):
    print(result, type(result))
````

## Exercise 3: String basics and checks

Use this text:

```python
text = "Python makes practice easy"
```

Tasks:
1. Print the first character and the last character.
2. Print the length of `text`.
3. Check if `"practice"` is in `text`.
4. Check if `"hard"` is not in `text`.
5. Use an `if` statement to print a message when `"Python"` is in `text`.

````python
text = "Python makes practice easy"
print('First:', text[0])
print('Last:', text[-1])
print('Length:', len(text))
print('Contains practice:', 'practice' in text)
print('Does not contain hard:', 'hard' not in text)
if 'Python' in text:
    print('The sentence mentions Python.')
````

## Exercise 4: Slicing, modifying, concatenation

Use:

```python
word = "Programming"
first = "Good"
second = "Morning"
```

Tasks:
1. Slice and print `gram` from `word`.
2. Print every second character of `word`.
3. Print `word` in uppercase and lowercase.
4. Replace `"Pro"` with `"Re"` and print result.
5. Concatenate `first` and `second` with a space.

````python
word = "Programming"
first = "Good"
second = "Morning"
print(word[3:7])  # gram
print(word[::2])
print(word.upper())
print(word.lower())
print(word.replace('Pro', 'Re'))
print(first + ' ' + second)
````

## Exercise 5: Formatting and escape characters

1. Create variables `item = "notebook"` and `price = 12.5`.
2. Use an f-string to print: `The notebook costs 12.50 dollars`.
3. Use `.format()` to print the same message.
4. Print a sentence with escaped double quotes around one word.
5. Print text that includes a new line (`\n`) and a tab (`\t`).

````python
item = "notebook"
price = 12.5
print(f'The {item} costs {price:.2f} dollars')
print('The {} costs {:.2f} dollars'.format(item, price))
print("The word \"Python\" is quoted.")
print('First line\n\tIndented second line')
````
