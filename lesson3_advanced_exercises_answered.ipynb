## Exercise 1: Truth table mini-lab

- Create booleans `p = True`, `q = False`.
- Print results for: `p and q`, `p or q`, `not p`, `not q`.
- Add one compound expression with parentheses and print its result.

```python
p = True
q = False
print('p and q:', p and q)
print('p or q:', p or q)
print('not p:', not p)
print('not q:', not q)
print('(p or q) and (not q):', (p or q) and (not q))
```

## Exercise 2: Comparison chain

- Given `a = 12`, `b = 12`, `c = 7`, print at least 6 comparison results.

## Exercise 3: Operator precedence check

- Create a complex boolean expression using `and`, `or`, and `not`.
- Evaluate it once without extra parentheses and once with explicit parentheses.
- Print both and comment whether they match.

```python
p = True
q = False
r = True
without_extra_parentheses = p or q and not r
with_parentheses = p or (q and (not r))
print('Without extra parentheses:', without_extra_parentheses)
print('With explicit parentheses:', with_parentheses)
print('Results match:', without_extra_parentheses == with_parentheses)
# They match because Python evaluates not before and, and and before or.
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]
print('Equal contents:', list1 == list2)
print('Same object:', list1 is list2)
alias = list1
print('Alias equal:', list1 == alias)
print('Alias same object:', list1 is alias)
# == compares contents; is checks whether variables refer to the same object.
# list1 and list2 are distinct lists, while alias refers to list1.
```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
items = ["python", "java", "go", "rust"]
queries = ["python", "javascript", "go", "ruby"]
for query in queries:
    if query in items:
        print(f'{query}: found')
    elif query not in items:
        print(f'{query}: not found')
```
